```python
from databricks.sdk import WorkspaceClient
import getpass

w = WorkspaceClient()

# Primero crea el alcance
try:
    w.secrets.create_scope(scope="consultech")
    print(" Alcance 'consultech' creado correctamente")
except Exception as e:
    print(f"ℹ Ya existe o error: {str(e)}")

# Ahora agrega tus claves (copia y pega cada una cuando te lo pida)
groq_key = getpass.getpass("Ingresa el valor de groq-key: ")
w.secrets.put_secret(scope="consultech", key="groq-key", string_value=groq_key)

serper_key = getpass.getpass("Ingresa el valor de serper-key: ")
w.secrets.put_secret(scope="consultech", key="serper-key", string_value=serper_key)

print(" Claves guardadas en el alcance")
```
